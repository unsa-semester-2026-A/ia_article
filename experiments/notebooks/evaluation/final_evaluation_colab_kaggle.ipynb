# Evaluación final configurable

Este orquestador instala el paquete editable, valida las pruebas y ejecuta todas las condiciones con pesos disponibles en `final_evaluation_config.json`. Cada condición publica únicamente un ZIP reproducible en su carpeta Drive.

```python
import os
import subprocess
import sys
from pathlib import Path

WORK_ROOT = Path('/kaggle/working') if Path('/kaggle/working').exists() else Path('/content')
EXPERIMENTS_DIR = Path(os.environ.get('EXPERIMENTS_DIR', WORK_ROOT / 'ia_article' / 'experiments'))
CONFIG_PATH = Path(os.environ.get('FINAL_EVALUATION_CONFIG', WORK_ROOT / 'final_evaluation_config.json'))
assert EXPERIMENTS_DIR.joinpath('pyproject.toml').is_file(), EXPERIMENTS_DIR
assert CONFIG_PATH.is_file(), f'Config inexistente: {CONFIG_PATH}'
subprocess.run([sys.executable, '-m', 'pip', 'install', '-q', '-e', '.[cloud]'], cwd=EXPERIMENTS_DIR, check=True)
subprocess.run([sys.executable, '-m', 'pytest', 'src/evaluation/', '-q'], cwd=EXPERIMENTS_DIR, check=True)
print('Entorno y pruebas listos:', EXPERIMENTS_DIR)
```

```python
subprocess.run(
    [sys.executable, '-m', 'src.evaluation.final_runner', '--config', str(CONFIG_PATH)],
    cwd=EXPERIMENTS_DIR,
    check=True,
)
```
