```python
# 4. Prepare dataset configuration file
import yaml

template_path = 'ia_article/experiments/configs/smart_dataset.yaml'
target_path = '/kaggle/working/smart_dataset.yaml'

# Ensure dataset path matches the Kaggle dataset mount point
# Typically /kaggle/input/smart-dataset/dataset or /kaggle/input/smart-original/dataset
kaggle_dataset_path = '/kaggle/input/smart-dataset/dataset'

if os.path.exists(template_path):
    with open(template_path, 'r') as f:
        data_config = yaml.safe_load(f)
    
    data_config['path'] = kaggle_dataset_path
    
    with open(target_path, 'w') as f:
        yaml.dump(data_config, f, default_flow_style=False)
    print(f"✓ Created dataset config at {target_path} pointing to {kaggle_dataset_path}")
else:
    print(f"ERROR: Template config not found at {template_path}")
```
